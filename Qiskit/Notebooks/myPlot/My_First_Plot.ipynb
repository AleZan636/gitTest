My First Plot. The first example of an application implemented via Jupyter Notebook : the 'python' implementation of 'y = sinx + cosx' function.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 400)
y = np.sin(x) + np.cos(x)
```

```python
plt.figure(figsize=(8,4))
plt.plot(x, y, label='sin(x) + cos(x)')
plt.title('My First Plot')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.show()           
```
